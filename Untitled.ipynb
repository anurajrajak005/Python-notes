```python
#check the number is prime or not
#ex 5 prime
#ex 6 not prime
num = int(input("Enter the number"))
count=0
for i in range(1,num+1):
  if num%i==0:
    count=count+1

if count==2:
  print("Prime")
else:
  print("not prime")
```

```python
#count the frequency of each element in a list
li = [1,3,54,1,54,2,42,2,1,2,23,2,1,3]
data = {}
for i in li:
  if i not in data:
    data[i]=1
  else:
    data[i]+=1


print(data)
```

```python
li=[1,3,5,52,5,2,3]
target = 10
ans=False

for i in range(len(li)):
  for j in range(i+1,len(li)):
    if li[i] + li[j] == target:
            ans = True
            break  # Stops the inner loop if a match is found

print(ans) 
```

```python
#Arrange list in assending order without using predefine function
li = [8,5,7,3,5,7,9,4,2,1]

for i in range(len(li)-1):
    for j in range(i+1,len(li)):
        if li[j]>li[i]:
           temp=li[i]
           li[i]=li[j]
           li[j]=temp
print(li)
```

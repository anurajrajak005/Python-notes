### Python support 3 categories of data types 
##### 1. *Basic Types (Integer , Float , Complex , Boolean  and string).*
##### 2. *container Types (List , Tuples , Sets and Dictionary).*
##### 3. *user-defind Class.* 

#### What is a Data Type in Python?
###### •	*A data type tells what kind of value a variable can hold.*
###### •	*In simple words → It defines the type of data stored in a variable (numbers, text, true/false, etc.).*
###### •	*Python is a dynamically typed language → You don’t need to declare data types; Python decides automatically based on the value.*

- #### 1. *integer*

### Container Types Function 

- *List*

```python
print([1,2,3,3,4,4,4,5,5,6,6,67,7])
```

- *Tuple*

```python
print(1,2,3,4,5,56,6,)
```

```python
print((1,2,3,4,5,56,6,))
```

- *Sets*

```python
print({1,23,33,99,378,9})
```

- *Dict*

```python
print({"Nitesh", 30, "male"})
```

```python
print({'Name':"Nitesh", "Age":30 , "Gender":'male'})
```

```python
print({
    "name":"Anuraj Rajak",
    "Age":21,
    "Gender": "Male",
    "Contract": 9669539437,
    "City": "Bhopal",
    "States": "MP"
})
```

```python
# Write a lambda function to sort a list of tuples based on the second element.
# Input: [(2,5), (1,2), (4,1)] → Output: [(4,1), (1,2), (2,5)]
data = [(2,5), (1,2), (4,1)]
print(sorted(data,key= lambda x:x[1]))
```
